```python
import math
import heapq
from typing import Dict, List, Tuple, Optional

class Node:
    def __init__(self, node_id: str, parent: Optional['Node'] = None, g: float = 0.0, h: float = 0.0):
        self.node_id = node_id
        self.parent = parent
        self.g = g
        self.h = h
        self.f = g + h

    def __lt__(self, other: 'Node') -> bool:
        return self.f < other.f

def calculate_heuristic(node_id: str, goal_id: str, coords: Dict[str, Tuple[float, float]]) -> float:
    x1, y1 = coords[node_id]
    x2, y2 = coords[goal_id]
    return math.sqrt((x1 - x2) ** 2 + (y1 - y2) ** 2)


def load_graph_from_file(file_path: str = "input.txt"):
    coords: Dict[str, Tuple[float, float]] = {}
    adj_list: Dict[str, List[Tuple[str, float]]] = {}

    with open(file_path, "r") as file:
        lines = [line.strip() for line in file if line.strip()]

    line_idx = 0
    
  
    if lines[line_idx].startswith('V'):
        line_idx += 1

    num_vertices = int(lines[line_idx])
    line_idx += 1

    for _ in range(num_vertices):
        parts = lines[line_idx].split()
        node_id = parts[0]
        x_val = float(parts[1])
        y_val = float(parts[2])
        coords[node_id] = (x_val, y_val)
        adj_list[node_id] = []
        line_idx += 1

   
    if line_idx < len(lines) and lines[line_idx].startswith('E'):
        line_idx += 1

    num_edges = int(lines[line_idx])
    line_idx += 1

    for _ in range(num_edges):
        parts = lines[line_idx].split()
        u_node = parts[0]
        v_node = parts[1]
        cost_val = float(parts[2])
        adj_list[u_node].append((v_node, cost_val))
        line_idx += 1

    start_node = lines[line_idx]
    line_idx += 1
    goal_node = lines[line_idx]

    return coords, adj_list, start_node, goal_node

def astar_search(
    coords: Dict[str, Tuple[float, float]],
    adj_list: Dict[str, List[Tuple[str, float]]],
    start_node: str,
    goal_node: str
):
    start_h = calculate_heuristic(start_node, goal_node, coords)
    start_state = Node(node_id=start_node, parent=None, g=0.0, h=start_h)

    min_q = []
    heapq.heappush(min_q, start_state)

    while min_q:
        curr_state: Node = heapq.heappop(min_q)

        if curr_state.node_id == goal_node:
            path = []
            curr = curr_state
            while curr:
                path.append(curr.node_id)
                curr = curr.parent
            path.reverse()
            
            print(f"Solution path {'-'.join(path)}")
            cost_str = int(curr_state.g) if curr_state.g.is_integer() else curr_state.g
            print(f"Solution cost {cost_str}")
            return path, curr_state.g

        for neighbor, edge_cost in adj_list.get(curr_state.node_id, []):
            g_new = curr_state.g + edge_cost
            h_new = calculate_heuristic(neighbor, goal_node, coords)
            neighbor_state = Node(node_id=neighbor, parent=curr_state, g=g_new, h=h_new)
            heapq.heappush(min_q, neighbor_state)

    print("Path not found!")
    return None, float('inf')

if __name__ == "__main__":
    coords_data, adj_data, start, goal = load_graph_from_file("input.txt")
    astar_search(coords_data, adj_data, start, goal)
```
